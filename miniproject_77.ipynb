```python
import pandas as pd
import numpy as np

from sklearn.model_selection import train_test_split
from sklearn.feature_extraction.text import TfidfVectorizer
from sklearn.naive_bayes import MultinomialNB
from sklearn.metrics import accuracy_score, classification_report, confusion_matrix
```

```python
data = {
    "subject": [
        "Urgent meeting tomorrow",
        "Project deadline today",
        "Exam schedule released",
        "Important assignment submission",
        "Interview scheduled tomorrow",
        "Emergency server issue",
        
        "Team meeting reminder",
        "Project discussion",
        "Weekly progress update",
        "College event information",
        "Department meeting",
        "Project review meeting",
        
        "Weekend sale offer",
        "Newsletter subscription",
        "Happy birthday",
        "Daily promotional email",
        "Social media notification",
        "Advertisement offer"
    ],

    "body": [
        "Please attend the urgent meeting tomorrow morning.",
        "The project deadline is today. Please submit immediately.",
        "The examination schedule has been released. Check it now.",
        "Your assignment must be submitted before the deadline.",
        "Your job interview is scheduled for tomorrow.",
        "The server is down and requires immediate attention.",

        "This is a reminder about our team meeting tomorrow.",
        "Let us discuss the project progress this week.",
        "Here is the weekly progress report from our team.",
        "Information about the upcoming college event.",
        "The department meeting will be conducted this Friday.",
        "The project review meeting is scheduled next week.",

        "Get exciting discounts this weekend.",
        "Here is your weekly newsletter.",
        "Wishing you a very happy birthday.",
        "Check out our latest promotional offers.",
        "You have a new social media notification.",
        "Special advertisement offer available now."
    ],

    "priority": [
        "High",
        "High",
        "High",
        "High",
        "High",
        "High",

        "Medium",
        "Medium",
        "Medium",
        "Medium",
        "Medium",
        "Medium",

        "Low",
        "Low",
        "Low",
        "Low",
        "Low",
        "Low"
    ]
}

df = pd.DataFrame(data)

df
```

```python
df["text"] = df["subject"] + " " + df["body"]

df[["text", "priority"]].head()
```

```python
X = df["text"]
y = df["priority"]

X_train, X_test, y_train, y_test = train_test_split(
    X,
    y,
    test_size=0.3,
    random_state=42,
    stratify=y
)

print("Training samples:", len(X_train))
print("Testing samples:", len(X_test))
```

```python
tfidf = TfidfVectorizer(
    lowercase=True,
    stop_words="english"
)

X_train_tfidf = tfidf.fit_transform(X_train)
X_test_tfidf = tfidf.transform(X_test)

print("TF-IDF Training Shape:", X_train_tfidf.shape)
```

```python
model = MultinomialNB()

model.fit(X_train_tfidf, y_train)

print("Model training completed!")
```

```python
y_pred = model.predict(X_test_tfidf)

accuracy = accuracy_score(y_test, y_pred)

print("Accuracy:", accuracy)
```

```python
print(classification_report(y_test, y_pred))
```

```python
subject = "Urgent project submission"
body = "The project must be submitted today. Please complete it immediately."

new_email = subject + " " + body

new_email_tfidf = tfidf.transform([new_email])

prediction = model.predict(new_email_tfidf)

print("Predicted Priority:", prediction[0])
```
